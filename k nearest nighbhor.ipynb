# ****K-Nearest Neighbors (KNN) Regressor and Classification in Python****

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
df.head()
```

```python
# Split the data
x =df.drop('species',axis=1)
y =df['species']
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(x,y)
```

```python
new_measurementv = np.array([[2.5,1.5,4.4,6.2]])
knn.predict(new_measurementv)
```

```python
# train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
#fit the model
knn.fit(x_train,y_train)
```

```python
y_pred = knn.predict(x_test)
```

```python
#evaluate the model
from sklearn.metrics import classification_report,confusion_matrix,accuracy_score
print('confusion matrix:\n',confusion_matrix(y_test,y_pred)
      )
print('classification report:\n',classification_report(y_test,y_pred))
print('accuracy score:\n',accuracy_score(y_test,y_pred))
```

```python
sns.pairplot(df,hue='species',palette='Dark2')
```

```python
sns.heatmap(confusion_matrix(y_test,y_pred),annot=True,fmt='d',cmap='Blues')
```

```python
#reggression problem of tips dataset
df = sns.load_dataset('tips')
df.head()
```

```python
x = df.drop('tip',axis=1) #split the data into features and target variable
y = df['tip']
```

```python
from sklearn.preprocessing import LabelEncoder

for col in x.columns:
    if x[col].dtype == 'object' or x[col].dtype == 'category':
        le = LabelEncoder()
        x[col] = le.fit_transform(x[col])

print(x.head())
print(x.dtypes)   # sab numeric hone chahiye (int/float)
```

```python
#train test split
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor
x_train,x_test,y_train,y_test = train_test_split(x,y,
                                                 test_size=0.3)
model = KNeighborsRegressor(n_neighbors=5,metric='minkowski',p=2)

#fit the model
model.fit(x_train,y_train)


#y pridict
y_pred = model.predict(x_test)


#evalate the model
from sklearn.metrics import mean_squared_error,r2_score

print("Mean squared error:\n",mean_squared_error(y_test,y_pred)
      )
print('R2 Score:\n',r2_score(y_test,y_pred))
```
