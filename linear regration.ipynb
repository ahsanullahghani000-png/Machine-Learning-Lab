# *****Linear Regeression*****

```python
import pandas as pd
import numpy as np  
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,r2_score
```

```python
df = sns.load_dataset('tips')
```

```python
sns.scatterplot(x='total_bill',y='tip',data=df)
```

```python
x = df[['total_bill']]
y = df['tip']
```

```python
x_train,x_test,y_train,y_test  = train_test_split(x,y,train_size=0.8)
#Call the model
model = LinearRegression()
```

```python
#train the model
model.fit(x_train,y_train)
```

```python
print("Model Coefficient:",model.coef_)
print("Model Intercept:",model.intercept_)
print("equation of line",model.intercept_,'+',model.coef_,'*x')
```

```python
#predict
y_pred = model.predict(x_test)
```

```python
#evaluate the model
print("Mean Squared Error:",mean_squared_error(y_test,y_pred))
print("R2 Score:",r2_score(y_test,y_pred))
```

**Plot the data**

```python
plt.scatter(x_test,y_test)
plt.plot(x_test,y_pred,color='red')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Linear Regression')
plt.show()
```

```python
#now pridict the tip for a total bill of 100
new_bill = np.array([[100]])
predicted_tip = model.predict(new_bill)
print("Predicted tip for a total bill of 100 is:",predicted_tip[0])
```
