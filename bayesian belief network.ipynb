# **Bayesian Belief Network**

```python
# Simple Bayesian Belief Network (bina kisi library ke)
#
# Network:   Rain ---> Sprinkler
#              \           |
#               \          v
#                ------> Grass
#
# Har variable: 1 = Yes (Haan), 0 = No (Nahi)

# ---------- Step 1: CPTs (probability tables) ----------

# P(Rain = 1)
P_rain = 0.2

# P(Sprinkler = 1 | Rain)
P_sprinkler = {
    1: 0.01,   # Barish hui to sprinkler chalne ka chance
    0: 0.40,   # Barish nahi hui to sprinkler chalne ka chance
}

# P(Grass = 1 | Rain, Sprinkler)    key = (Rain, Sprinkler)
P_grass = {
    (1, 1): 0.99,
    (1, 0): 0.80,
    (0, 1): 0.90,
    (0, 0): 0.00,
}


# ---------- Step 2: Joint probability ----------
# P(R, S, G) = P(R) * P(S|R) * P(G|R,S)

def joint(r, s, g):
    p_r = P_rain if r == 1 else 1 - P_rain

    p_s = P_sprinkler[r]
    p_s = p_s if s == 1 else 1 - p_s

    p_g = P_grass[(r, s)]
    p_g = p_g if g == 1 else 1 - p_g

    return p_r * p_s * p_g


# ---------- Step 3: Inference (sawal ka jawab) ----------

def query(target, evidence):
    """
    target   : 'R', 'S' ya 'G'  (jis ki probability chahiye)
    evidence : dict, jaise {'G': 1}  (jo pata hai)
    Return   : P(target = 1 | evidence)
    """
    total = 0.0          # P(evidence)
    target_yes = 0.0     # P(target = 1, evidence)

    for r in (0, 1):
        for s in (0, 1):
            for g in (0, 1):
                values = {"R": r, "S": s, "G": g}

                # Agar ye combination evidence se match nahi karti to skip
                if any(values[k] != v for k, v in evidence.items()):
                    continue

                p = joint(r, s, g)
                total += p
                if values[target] == 1:
                    target_yes += p

    return target_yes / total


# ---------- Step 4: Run karo ----------

print("Joint probability:")
print("P(Rain=0, Sprinkler=1, Grass=1) =", round(joint(0, 1, 1), 4))

print("\nInference (Ghaas geeli hai, to?):")
print("P(Rain = Yes      | Grass = Wet) =", round(query("R", {"G": 1}), 4))
print("P(Sprinkler = Yes | Grass = Wet) =", round(query("S", {"G": 1}), 4))

print("\nAgar ghaas geeli hai aur barish nahi hui:")
print("P(Sprinkler = Yes | Grass = Wet, Rain = No) =",
      round(query("S", {"G": 1, "R": 0}), 4))
```
