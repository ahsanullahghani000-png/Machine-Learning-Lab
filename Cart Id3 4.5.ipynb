# ID3 (Iterative Dichotomiser 3) Algorithm

## Introduction

ID3 is a Decision Tree algorithm used for classification. It selects the best attribute using **Information Gain** to split the dataset into smaller groups.

## Basic Concept

* Entropy measures the impurity or uncertainty in a dataset.
* Information Gain measures how much uncertainty is reduced after splitting the dataset.
* ID3 selects the attribute with the highest Information Gain.

## Algorithm Steps

1. Start with the complete training dataset.
2. Calculate the entropy of the dataset.
3. Calculate Information Gain for each attribute.
4. Select the attribute with the highest Information Gain.
5. Split the dataset based on the selected attribute.
6. Repeat the process until a stopping condition is reached.
7. Return the final Decision Tree.

## Advantages

* Simple and easy to understand.
* Useful for classification problems.
* Uses Information Gain for attribute selection.

## Limitations

* Can favor attributes with many unique values.
* The original ID3 algorithm mainly works with categorical attributes.
* It can overfit the training data.

## Conclusion

ID3 builds a Decision Tree by selecting the attribute that provides the highest Information Gain at each step.


```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
from sklearn.tree import DecisionTreeClassifier, plot_tree

# Load the complete Titanic dataset
df = sns.load_dataset("titanic")

# Display dataset information
print("Dataset Shape:", df.shape)
print(df.head())
print(df.isnull().sum())

# Select useful columns
df = df[
    ["survived", "pclass", "sex", "age",
     "sibsp", "parch", "fare", "embarked"]
]

# Handle missing values
df = df.dropna()

# Separate features and target
X = df.drop("survived", axis=1)
y = df["survived"]

# Convert categorical columns into numerical columns
X = pd.get_dummies(
    X,
    columns=["sex", "embarked"],
    drop_first=True,
    dtype=int
)

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```
