<p style="

The <b style="color:#00f2fe;">Find-S Algorithm</b> is a simple Machine Learning algorithm
used to find a pattern from a set of training examples.
</p>

<p style="
    color: #e5faff;
    font-family: 'Segoe UI', system-ui, sans-serif;
    font-size: 17px;
    line-height: 1.8;
">
It starts with the <b style="color:#00f2fe;">most specific hypothesis</b>
and then checks the training examples one by one.
</p>

<p style="
    color: #e5faff;
    font-family: 'Segoe UI', system-ui, sans-serif;
    font-size: 17px;
    line-height: 1.8;
">
The algorithm looks only at <b style="color:#22c55e;">positive examples (Yes)</b>.
When two positive examples have different values for an attribute,
that attribute is changed to <b style="color:#facc15;">?</b>.
</p>

<p style="

">

<span style="color:#00f2fe;"> → </span>

Compare
<span style="color:#00f2fe;"> → </span>

<span style="color:#00f2fe;"> → </span>

<p style="

<b style="color:#00f2fe;">In simple words:</b>
Find-S learns the common features of positive examples and creates
the most specific rule that matches them.
</p>

</div>


```python
# Find-S Algorithm

data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

# Initial hypothesis
hypothesis = ['0', '0', '0', '0', '0', '0']

# Find-S Algorithm
for row in data:

    # Only positive examples
    if row[-1] == 'Yes':

        for i in range(6):

            if hypothesis[i] == '0':
                hypothesis[i] = row[i]

            elif hypothesis[i] != row[i]:
                hypothesis[i] = '?'

# Final result
print("Final Hypothesis:", hypothesis)
```
